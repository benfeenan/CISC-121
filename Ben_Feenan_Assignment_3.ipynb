#Stack Problem: Daily Temperatures


#1. Understanding
To solve this question I will use a stack. I will create an array to store the ammount of time waited. The current day will be the top of my stack. I will compare the next temperature to the temperature at the top of my stack. If it is warmer I will pop the top value, check current - pop, push the value of my current, and that will be how many days it waited. If it is not warmer, I will push the current value which will become my new top. I will move to the next day and compare it with my top value. I will continue this until I find a higher temperature then check it against the top until I get all the way back to the day I was originally checking. If I do not find a higher temperature nothing will happen and the array will remain 0.

# 2 Peer Review Code
After viewing https://leetcode.com/problems/daily-temperatures/ I can see that my logic of using a stack and popping the top value to figure out the amount of days waited. In this solution instead of going left to right they went right to left showing me that I can use stack logic in both directions to get the same result.

```python
#Draft Code
class Solution(object):
    def dailyTemperatures(self, temperatures):
        """
        :type temperatures: List[int]
        :rtype: List[int]
        """
        ans = [0]*len(temperatures) #create array to store wait time
        wait_time = []

        for current in range(len(temperatures)):
            while wait_time and wait_time[-1] < temperatures[current]:
                prev_day = wait_time.pop()
                ans[prev_day] = prev_day - current
            wait_time.append(current)
        return ans
```

# Recursion Climbing Stairs
To solve this question I need to print every possible combination of getting to the top of the stairs using steps of 1 or 2. To do this I will use backtracking recursion.

#1. Understanding
To solve this question I need to print every possible combination of getting to the top of the stairs using steps of 1 or 2. To do this I will use backtracking recursion. I will call my recursive function until each branch has reached the base case of n. If the current branch tries to go above n it will fail. I will count the number of valid branches.

#2.Peer Review Code
After viewing https://leetcode.com/problems/climbing-stairs/solutions/7352365/best-explaination-recursion-flowchart-by-ho58/ I realized that you could solve this question without having to generate all responses. The number is always dependant on the two numbers before it. My recursive function needs to be able to get to those previous numbers.

```python
#Draft Code
class Solution(object):
    def climbStairs(self, n):
        """
        :type n: int
        :rtype: int
        """
        def up_stairs(n):
            if n == 1:
                return 1
            else:
                return up_stairs(n-1) + up_stairs(n-2)
        return up_stairs(n)
```

#3. Final Correct Code


```python
class Solution(object):
    def climbStairs(self, n):
        """
        :type n: int
        :rtype: int
        """
        memo = {}
        def up_stairs(n):
            if n <= 2:
                return n
            if n in memo:
                return memo[n]

            memo[n] = up_stairs(n-1) + up_stairs(n-2)
            return memo[n]

        return up_stairs(n)
```
