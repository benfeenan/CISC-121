# Reverse Integer

1. I need to reverse the order of an integer while keeping its sign the same. To do this I will check whether x is positive or negative. If it is positive I can reverse the integer the same way as the palindrome then turn it back to an int. If it is positive I will reverse the string up to the first digit which will be the - sign. Then I will confirm it is within the range.

2. After viewing https://leetcode.com/problems/reverse-integer/solutions/6354708/elegant-python3-solution-with-detailed-explanation/ I see that I can use abs to get the absolute power of the number instead of trying to ignore the minus sine in the string.

```python
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        if x > 0:
            positive_int = int(str(x)[::-1])
            if -(2**31) < positive_int < (2**31):
                return(False)

        elif x < 0:
            negative_int = int(str(abs(x))[::-1])* -1
            if -(2**31) < negative_int < (2**31):
                return(False)
```

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """

        for i in range(len(nums)):
            for j in range(i,len(nums)):
                if i != j and i + j == target
                return [i,j]
```
